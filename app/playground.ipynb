```python
from langchain_core.tools import tool



@tool
def echo(text: str) -> str:
    """Echoes the input text back."""
    return f"Echo result: {text}"
```

```python
llm = ChatGroq(
    model="qwen/qwen3-32b"
).bind_tools([echo])

resp = await llm.ainvoke(
    {"use echo tool to echo hello"}
)

print(resp.tool_calls)
```
